```python
import random
def get_matrix_input():
    print("--- Distance Matrix Configuration ---")
    try:
        n = int(input("Enter the number of cities: "))
        if n < 3:
            print("❌ You need at least 3 cities to create a valid round-trip route!")
            return None
       
        print(f"\nEnter the distances row by row (space-separated numbers).")
        print("Note: The distance from a city to itself should be 0.")
       
        matrix = []
        for i in range(n):
            while True:
                row_str = input(f"Enter distances from City {i+1} to all {n} cities: ")
                row = [int(x) for x in row_str.split()]
               
                if len(row) != n:
                    print(f"⚠️ Error: You must enter exactly {n} values. Try again.")
                    continue
               
                matrix.append(row)
                break
        return matrix
    except ValueError:
        print("❌ Invalid input. Please enter integers only.")
        return None

# --- Get inputs from user ---
dist = get_matrix_input()

if dist:
    # Set algorithm constants
    n = len(dist)
    population_size = 10
    generations = 100
    mutation_rate = 0.1

    # Create initial population
    def create_route():
        return random.sample(range(n), n)

    population = [create_route() for _ in range(population_size)]

    # Calculate total distance of a route
    def fitness(route):
        total = 0
        for i in range(n):
            total += dist[route[i]][route[(i + 1) % n]]
        return total

    # Select the best route
    def selection(population):
        return min(population, key=fitness)

    # Crossover two parents
    def crossover(p1, p2):
        point = random.randint(1, n - 1)
        child = p1[:point]
        for city in p2:
            if city not in child:
                child.append(city)
        return child

    # Mutate a route
    def mutation(route):
        if random.random() < mutation_rate:
            i, j = random.sample(range(n), 2)
            route[i], route[j] = route[j], route[i]
        return route

    # Genetic Algorithm Loop
    for _ in range(generations):
        new_population = []

        for _ in range(population_size):
            p1 = selection(population)
            p2 = random.choice(population)

            child = crossover(p1, p2)
            child = mutation(child)

            new_population.append(child)

        population = new_population

    # Display best route and human-readable path formatting
    best = min(population, key=fitness)
    formatted_route = [city + 1 for city in best] + [best[0] + 1]
   
    # Generate visual path map string
    path_visual = " -> ".join(map(str, formatted_route))

    print("\n================= FINAL RESULTS =================")
    print(f"Best Route Order : {formatted_route}")
    print(f"Visual Travel Path: {path_visual}")
    print(f"Minimum Distance  : {fitness(best)} miles/km")
    print("=================================================")
```
